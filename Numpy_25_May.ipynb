### Array Manipulation

- np.concatenate
- np.hstack
- np.vstack
- np.column_stack
- np.hsplit
- np.vsplit

```python
arr4 = np.random.randint(30, 54, size = (5,5))
arr5 = np.random.randint(30, 54, size = (5,5))
```

```python
np.concatenate((arr4, arr5), axis = 0)  #concatenate the arrays in vertical direction
```

```python
np.concatenate((arr4, arr5), axis = 1)  #concatenate the arrays in the horizontal direction
```

```python
np.vstack((arr4, arr5)) #similar to concatenate with axis = 0
```

```python
np.hstack((arr4, arr5)) #similar to concatenate with axis = 1
```

```python
arr_big = np.random.randint(40, 81, size = (12,10))
```

```python
arr_big
```

```python
arr11, arr12 = np.hsplit(arr_big, 2)
```

```python
np.vsplit(arr_big, 2)
```
