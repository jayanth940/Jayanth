## **Classes and Objects**

The Car class serves as a blueprint or template for creating individual car objects. It defines the structure and behavior that all car objects will have. Here's how the class and objects relate:
1. `Class as a Template`: The Car class defines the attributes (make, model, year, color) and methods (startEngine, accelerate, brake) that any car object will possess.
2. `Objects as Instances`: We can create multiple car objects based on this class, each representing a specific car.
3. `Unique State`: Each car object has its own set of attribute values, representing its unique state. The honda and toyota objects have different makes, models, years, and colors.
4. `Shared Behavior`: All car objects share the same methods defined in the class. Both honda and toyota can call startEngine(), accelerate(), and brake().
5. `Memory Allocation`: When we declare the Car class, no memory is allocated. Memory is only allocated when we create car objects using the new keyword.
6. `Reusability`: The Car class can be used to create any number of car objects, promoting code reuse and organization.

### **Methods (User Defined Functions inside the classes)**

```python
class Account:

  def __init__(self, customername, accountnumber, balance):
    self.name = customername
    self.acc_no = accountnumber
    self.bal = balance

  def deposit(self, amount):
    if amount > 0:
      self.bal = self.bal + amount
    else:
      print('Deposit amount must be positive and greater than 0.')

  def withdraw(self, amount):
    if amount > 0 and amount <= self.bal:
      self.bal = self.bal - amount
    elif amount > self.bal:
      print('Insufficient balance.')
    else:
      print('Withdraw amount must be positive and greater than 0.')

  def __str__(self):
    return f'The account number {self.acc_no} belongs to {self.name} and the balance is INR {self.bal}.'
```

##### **1. Single Inheritance**

```python
class Person:

  def __init__(self, name, id):
    self.name = name
    self.id = id

  def get_details(self):
    return f'Name : {self.name}, ID : {self.id}'

class Student(Person):

  def __init__(self, name, id, grade, course):
    super().__init__(name, id)
    self.grade = grade
    self.course = course
```

```python
st1 = Student('Sam', 1234, 'A+', 'Computer Science')
```

```python
st1.name
```

```python
class Account:

  def __init__(self, customername, accountnumber, balance):
    self.name = customername
    self.acc_no = accountnumber
    self.bal = balance
    self.numtrans = 0
    self.maxtrans = 2

  def deposit(self, amount):
    if amount > 0 and self.numtrans < self.maxtrans:
      self.bal = self.bal + amount
      self.numtrans = self.numtrans + 1   #self.numtrans += 1
    elif self.numtrans >= self.maxtrans:
      print('Maximum number of transactions reached.')
    else:
      print('Deposit amount must be positive and greater than 0.')

  def withdraw(self, amount):
    if amount > 0 and amount <= self.bal and self.numtrans < self.maxtrans:
      self.bal = self.bal - amount
      self.numtrans = self.numtrans + 1   #self.numtrans += 1
    elif amount > self.bal:
      print('Insufficient balance.')
    else:
      print('Withdraw amount must be positive and greater than 0.')

  def __str__(self):
    return f'The account number {self.acc_no} belongs to {self.name} and the balance is INR {self.bal}.'

class SavingsAccount(Account):
  pass

class CurrentAccount(Account):
  def __init__(self, customername, accountnumber, balance, gender):
    super().__init__(customername, accountnumber, balance)  #it tells python to give access to parent init also
    self.maxtrans = 5
    self.gender = gender
```

##### **Let's implement the encapsulation on our banking application**

```python
class Account:

  def __init__(self, customername, accountnumber, balance):
    self.name = customername
    self.acc_no = accountnumber
    self.__bal = balance

  def deposit(self, amount):
    if amount > 0:
      self.__bal = self.__bal + amount
    else:
      print('Deposit amount must be positive and greater than 0.')

  def withdraw(self, amount):
    if amount > 0 and amount <= self.__bal:
      self.__bal = self.__bal - amount
    elif amount > self.__bal:
      print('Insufficient balance.')
    else:
      print('Withdraw amount must be positive and greater than 0.')

  def get_balance(self):  #this is a getter method
    return self.__bal

  def set_balance(self, newamount):   #this is a setter method
    if type(newamount) in [int, float]:
      self.__bal = self.__bal + newamount
    else:
      print('I will fire you if you even try to so something like this.')

  def __str__(self):
    return f'The account number {self.acc_no} belongs to {self.name} and the balance is INR {self.__bal}.'
```

```python
acc1 = Account('Virat', 100, 25000)
```

```python
acc1.get_balance()
```

```python
acc1.set_balance('Python')
```

```python
acc1.get_balance()
```

#### **Method Overloading**

```python
class Calculator:
  def add(self, a, b):
    return a + b

  def add(self, a, b, c):
    return a + b + c
```

```python
calc = Calculator()
```

```python
calc.add(3, 4)
```

```python
class Calculator:

  def add(self, a, b=0, c=0, d = 0):
    return a + b + c + d
```

```python
calc = Calculator()
```
