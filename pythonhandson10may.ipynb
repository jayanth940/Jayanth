```python
# Write the code to check whether the number is odd or even using function

def is_even(number):
  if number % 2==0:
    return "even"
  else:
    return "odd"

print(is_even(3))
```

```python
# input marks and print the grade based on the condition:
# 90-100 : A
# 80-89: B
# 70-79: C
# 60-69: D
# <60: F

marks = int(input("Enter the marks: "))

if marks >=90 and marks <=100:
  print("A")
elif marks>=80:
  print("B")
elif marks>=70:
  print("C")
elif marks>=60:
  print("D")
else:
  print("F")
```

```python
# check whether the number is prime

def is_prime(n):
  if n<=1:
    return False
  for i in range(2,n):
    if n%i==0:
      return False
  return True

print(is_prime(7))
```
