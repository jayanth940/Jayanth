```python
# What is OOPS?
# OOPS stands for object oriented programming
# It is programming style where we can organize the code using objects, which are real world entities like (car,student, employee details)
# Each object has:
# Data(called attributes or properties)
# Functions(called method) that operates on the data.
```

```python
# Why do we need oops?
# Organize code: It makes the code look cleaner and easier to manage.
# Reusability: We can resue the code using classes and objects
# Scalability: It's easier to add features without breaking the existing code.
```

```python
# When do we use oops?
# when building large or complex projects at that time we need oops.
# whenever we are working in a team project at that time we need oops.
```

```python
class calculator:
  def __init__(self,num1,num2):   # __init__ is is constructor
    self.num1=num1
    self.num2=num2
  def addition(self):
    return self.num1+self.num2
  def subtraction(self):
    return self.num1-self.num2
  def multiplication(self):
    return self.num1*self.num2
  def division(self):
    return self.num1/self.num2
```

```python
obj = calculator(12,6)
```

```python
obj.addition()
```

```python
obj.subtraction()
```

```python
obj.multiplication()
```

```python
obj.division()
```

```python
# Simple interest and compound interest using constructor
# SI = PRT/100
# CI = (P*(1+(R/100))**T)-P

class Interest:
  def __init__(self,P,R,T):
    self.P = P
    self.R = R
    self.T= T
  def Simple(self):
    return (self.P*self.R*self.T)/100
  def Compound(self):
    return (self.P*(1+(self.R/100))**self.T)-self.P
```

```python
i = Interest(1000,5,2)
```

```python
i.Simple()
```

```python
i.Compound()
```

```python
# Pillars
# Pillars for OOPS are the 4 main concepts that define how my object oriented programming works.
# there are 4 pillars: Inheritance, encapsulation, polymorphism and abstraction
```

```python
# Inheritance: One class(child) can use features from parent class.

# Types:
# Single: 1 parent class and 1 child class
# Multiple: many parents class and 1 child class
# Multilevel: level of inheritance in form of chaining G->P->C
# Hierarchical:  1 parent class and many child class
# Hybrid: mix of 2 or more types of inheritance.
```

```python
# single inheritance

class Parent:
  def hello(self):
    print("hello world")

class Child(Parent):
  def hi(self):
    print("how are you?")
```

```python
c = Child()
```

```python
c.hi()
```

```python
c.hello()
```

```python
# multiple inhertance

class Parent1:
  def __init__(self,name,age,rollno):
    self.name=name
    self.age=age
    self.rollno=rollno
  def display_details(self):
    print("name of the student: ", self.name)
    print("age of the student: ", self.age)
    print("rollno of the student: ", self.rollno)
```

```python
class Parent2:
  def __init__(self,marks):
    self.marks=marks
  def percentage(self):
    return (sum(self.marks)/len(self.marks))
```

```python
class Child(Parent1, Parent2):
  def __init__(self,name,age,rollno,marks,Class):
    self.Class=Class
    Parent1.__init__(self,name,age,rollno)
    #super().__init__(name,age,rollno)   # super() in python is used to give access to methods and properties of the parent to the child class
    Parent2.__init__(self,marks)

  def show_all(self):
    super().display_details()
    print(f"Percentage of the student:{self.percentage()}")
    print("Student is of class :", self.Class)

    if self.percentage()>=90 and self.percentage()<=100:
      print("grade A")
    elif self.percentage()>=80:
      print("grade B")
    elif self.percentage()>=70:
      print("grade C")
    else:
      print("grade D")
```

```python
obj = Child('devansh',14,67,[90,89,95,96,88],10)
```

```python
obj.display_details()
```

```python
obj.percentage()
```

```python
obj.show_all()
```

```python
# multiple

class Father:
  def show_father(self):
    print("this is father class")

class Mother:
  def show_mother(self):
    print("this is mother class")

class Child(Father,Mother):
  def show_child(self):
    print("this is our child class")

```

```python
c = Child()
```

```python
c.show_child()
```

```python
c.show_father()
```

```python
c.show_mother()
```
