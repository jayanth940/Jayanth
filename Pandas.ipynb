### **Data Selection using `iloc` and `loc`**

#### **iloc** : index based selection of the data

```python
market.head(15)
```

```python
market.iloc[5]
```

```python
market.iloc[[3, 65, 89, 769, 3421]]
```

```python
market.iloc[3, 4]
```

```python
market.iloc[100:501, [3,4,5]]
```

#### **loc** : label/text/name based data selection

```python
market.loc[100:501, ['Cust_id', 'Sales', 'Discount']]
```

### **Data Selection using conditions**

```python
market.head(15)
```

```python
market['Profit'] < 0   #mark True for the rows which has negative profit values
```

```python
market[market['Profit'] < 0]  #select all the rows which has negative profit values
```

```python
market['Sales'] > 100
```

```python
market[market['Sales'] > 100]
```

```python
(market['Profit'] < 0) & (market['Sales'] > 100)
```

```python
market[(market['Profit'] < 0) & (market['Sales'] > 100)]
```

```python
market.loc[(market['Profit'] < 0) & (market['Sales'] > 100), ['Cust_id', 'Sales', 'Discount']]
```

```python
market[(market['Profit'] < 0) | (market['Sales'] > 100)]
```

```python
and : &
or : |
not : ~
```

```python
market[~(market['Profit'] < 0)]
```
