```python
import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns
```

### **Types of Plots**

#### **1. Scatter Plot (Bivariate Plot)**

- Used to represent the relationship between 2 numerical columns.

```python
tips = pd.read_csv('tips.csv')
```

```python
tips.head()
```

```python
x = tips['total_bill']
y = tips['tip']

plt.scatter(x, y, color = 'green', marker = '*', s = 300, edgecolor = 'red')
plt.xlabel('Total Bill')
plt.ylabel('Tip')
plt.title('Total Bill vs Tip Paid by the Customers')
plt.xticks(range(0, 55, 5), rotation = 90)
plt.yticks(range(0, 12, 1))
plt.show()
```

#### **2. Histogram (Univariate Plot)**

- Reprsents the frquency of the data in some range. We use a numerical column to make a histogram.

- It also represents distribution of the data.

```python
tips.head()
```

```python
x = tips['total_bill']

plt.hist(x, bins = 20)
plt.xlabel('Total Bill')
plt.ylabel('Frequency')
plt.show()
```

#### **3. Bar Chart (Univariate Plot)**

- Unlike histogram, it is used to represent a categorical column.

```python
tips.head()
```

```python
tips['sex'].value_counts()
```

```python
plt.bar(tips['sex'].value_counts().index, height = tips['sex'].value_counts().values)
plt.show()
```

```python
tips['day'].value_counts()
```

```python
plt.bar(tips['day'].value_counts().index, height = tips['day'].value_counts().values, width = 0.7,
        color = ['green', 'red', 'yellow', 'pink'])
plt.xlabel('Days')
plt.ylabel('Number of Customers')
plt.title('Number of customers day-wise')
plt.xticks(rotation = 90)
plt.show()
```

#### **Figures and Subplots**

- If we want to make some plots side by side or along with each other we can use the concept of subplots for that.

```python
plt.figure(figsize = (12, 5))

plt.subplot(1,2,1)
plt.scatter(x, y, color = 'green', marker = '*', s = 300, edgecolor = 'red')
plt.xlabel('Total Bill')
plt.ylabel('Tip')
plt.title('Total Bill vs Tip Paid by the Customers')
plt.xticks(range(0, 55, 5), rotation = 90)
plt.yticks(range(0, 12, 1))


plt.subplot(1,2,2)
plt.bar(tips['day'].value_counts().index, height = tips['day'].value_counts().values, width = 0.7,
        color = ['green', 'red', 'yellow', 'pink'])
plt.xlabel('Days')
plt.ylabel('Number of Customers')
plt.title('Number of customers day-wise')
plt.xticks(rotation = 90)

plt.show()
```

### **Data Visualization using Seaborn**

#### **1. Scatterplot using Seaborn**

```python
tips.head()
```

```python
sns.set() #change the default style from matplotlib style to seaborn style
```

```python
sns.relplot(x = 'total_bill', y = 'tip', data = tips, kind = 'scatter', hue = 'sex', col = 'day', palette = ['green', 'orange'])
plt.savefig('1.png')
plt.show()
```

#### **2. Regplot (Regression Plot)**

```python
sns.regplot(x = 'total_bill', y = 'tip', data = tips)
plt.show()
```

### **Distribution Plots (Univariate Plots)**

1. Histogram
2. K.D.E plot(Kernel Density Estimate Plot)
3. Displot(Distribution Plot)

```python
sns.displot(x = 'total_bill', data = tips)
plt.show()
```

```python
sns.kdeplot(x = 'total_bill', data = tips)
plt.show()
```

#### **Countplot (Univariate Plot)**

- Alternate of Bar chart

```python
sns.countplot(x = 'sex', data = tips, hue = 'smoker')
plt.show()
```

#### **Pairplot (Multivariate Plot)**

- It is a special type of plot which is a collection of scatter plots.

```python
iris = pd.read_csv('iris.csv')
```

```python
iris.head()
```

```python
sns.pairplot(iris, hue = 'species')
plt.show()
```

#### **Boxplot**

- Boxplot has two versions : Univariate version and Bivariate version

```python
#univariate version of boxplot is used to detect the presence of outliers in the numerical data
sns.boxplot(x = 'total_bill', data = tips)
plt.show()
```

```python
tips.head()
```

```python
#bivariate version of boxplot is used to represent the relationship between a numerical vs a categorical column
sns.boxplot(x = 'sex', y= 'total_bill', data = tips)
plt.show()
```

#### **Line Plot (Bivariate Plot)**

- Used to represent a time series data. Which means we represent a relationship between one time related column with one numerical column.

```python
flights = pd.read_csv('flights.csv')
```

```python
flights.head()
```

```python
sns.relplot(x = 'year', y= 'passengers', data = flights, kind = 'line')
plt.show()
```
