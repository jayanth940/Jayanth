```python
# Slicing : It allows to extract a part of list , string, tuple using ranges on indexes
# syntax : variable[start: end-1: step]
# start : the index from where the slicing will start
# end : the index where the slicing will end/stop
# step : how many elements i want to skip

text = 'PYTHON'
       #012345
print(text[1:4]) #output : YTH
print(text[:3])  #output : PYT
print(text[2:])
print(text[::2]) #output :
print(text[::-1])
```

```python
numbers = [10,20,30,40,50,60]

print(numbers[1:4])   #output: 20,30,40     1. 2. 3. 4. 5.
print(numbers[:3])   #output: 10,20,30
print(numbers[3:])   # 40,50,60
print(numbers[::2])   # 10,30,50
print(numbers[::-1])   # 60,50,40,30,20,10
```

```python
# Control flow statements
# 1. Break : stop the loop
# 2. Continue : To skip the current iteration and move to next.
# 3. Pass: Does nothing, it will just pass ignoring the conditions.
```

```python
# Break

for i in range(5):   # i = 0,1,2,3,4
  if i == 2:
    break
  print(i)
```

```python
# Continue

for i in range(5):
  if i == 2:
    continue
  print(i)
```

```python
# pass

for i in range(5):
  if i == 2:
    pass
  print(i)
```

```python
# calculator

def add(a,b):
  return a+b

def subtract(a,b):
  return a-b

def multiply(a,b):
  return a*b

def divide(a,b):
  if b ==0:
    return
  return a/b


num1 = int(input("Enter the number 1 : "))
num2 = int(input("Enter the number 2 : "))

print("Choose the operation you want to perform : +,-,*,/")
operation = input("Enter the operator : ")

if operation == "+":
  print("Result : ", add(num1,num2))

elif operation == "-":
  print("Result : ", subtract(num1,num2))

elif operation == "*":
  print("Result : ", multiply(num1,num2))

elif operation == "/":
  print("Result : ", divide(num1,num2))

else:
  print("Invalid input")
```

```python
# Write the code for factorial number


def factorial(n):
  fact = 1
  for i in range(1,n+1):  #multiply fact by every number from 1 to n
    fact *= i
  return fact

num = int(input("Enter the number : "))
print("Factorial of ", num, "is", factorial(num))


```
