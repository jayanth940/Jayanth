# Constructor (`__init__()`)
* It initialize the attributes of a class whenever we create an object
* We can create the constructor by using` __init__ `method



---
**Note: whenever we create a function(method) inside a class, we always pass the first parameter as `self`**



self -> it is a connector that connects my object with the data memebers

```python
# Create the employee class
class Employee:
  def __init__(self,empName,empDomain,empSalary):
    self.empName = empName #self.empName --> this is my attribute
    self.empDomain = empDomain
    self.empSalary = empSalary
    print('constructor called')

  def display(self):
    print(f'Name : {self.empName}, Domain: {self.empDomain}, Salary: {self.empSalary} ')
```

# Inheritance
* Passing on the data members from parent class(Base class) to the child class(Derived class)

### Types of Inheritance
* Single level inheritance
* Multilevel inheritance
* Multiple inheritance
* hierarichal inheritance


---
# Single level inheritance
One parent class -----> One child class


```python
# Parent Class
class Parent:
  def Display(self):
    print('I am from Parent class')

# Child Class
class Child(Parent): # child class is inherted from parent class
  def Show(self):
    print('This is my child class')
```

```python
# Object
ob = Child()
ob.Show() # This is method belongs to child class only
ob.Display() # This is the method that belongs to Parent class
```

# Multi Level inheritance
Parent class -----> Child class ------> Grand Child class ----> go on

```python
# Parent Class
class Parent:
  def Display(self):
    print('I am from Parent class')

# Child Class
class Child(Parent): # child class is inherted from parent class
  def Show(self):
    print('This is my child class')

class GrandChild(Child):
  def output(self):
    print('Grand Child')
```

```python
g1 = GrandChild()
g1.output() #v3
g1.Show()   #v2
g1.Display()#v1
```

# Multiple Inheritance
Parent1 + Parent2 -----> Child class

```python
class Parent1:
  def Display(self):
    print('Parent1')

class Parent2:
  def Show(self):
    print('Parent2')

class Child(Parent1,Parent2):
  def output(self):
    print('Child')
```

```python
ob1 = Child()
ob1.Display()
ob1.Show()
ob1.output()
```

# hierarchical inheritance
Parent class -----> Child1 + Child2

```python
class Parent:
  def Display(self):
    print('I am Parent class')

class Child1(Parent):
  def Show(self):
    print('Child1')

class Child2(Parent):
  def output(self):
    print('Child2')
```

```python
ob1 = Child1()
ob1.Display()
ob1.Show()

ob2 = Child2()
ob2.Display()
ob2.output()
# ob2.Show()
```

# Polymorphism
Poly - many Morphism - Forms
* One method name having different functionalities in different classes

```python
class Square:
  def area(self):
    print('Area of Square')

class Rectangle(Square):
  def area(self):
    super().area() # my parent class area method is going to call here
    print('Area of Rectangle')
```

```python
ob = Square()
ob.area()
```

```python
ob2 = Rectangle()
ob2.area()
```

### For multiple parent class

```python
class Square:
  def area(self):
    print('Area of Square')

class Rectangle:
  def area(self):
    print('Area of Rectangle')

class Circle(Rectangle,Square):
  def area(self):
    super().area() # Super function will call the first parent class written
    super().area()
    print('Area of circle')
```

```python
ob = Circle()
ob.area()
```

```python
class Square:
  def area(self,a,b,c):
    print('Area of Square')

class Rectangle:
  def area(self):
    print('Area of Rectangle')

class Circle(Rectangle,Square):
  def area(self):
    Rectangle.area(self) # when we use this method we need to pass(self parameter)
    Square.area(self)
    print('Area of circle')
```

```python
ob = Circle()
# ob.area()
```

### Let's see polymorphism concept through constructor

```python
class Square:
  def __init__(self,side):
    self.side = side # Initializing my attribute

  def area(self):
    print(f'Area of Square is {self.side**2}')


class Rectangle(Square):
  def __init__(self,length,width,side):
    self.length = length
    self.width = width
    Square.__init__(self,side)

  def area(self):
    print(f'Area of rectangle is {self.length * self.width}')
    Square.area(self)
```

```python
ob = Rectangle(5,2,3)
ob.area()
```

# Abstraction
* Hiding the unnecessary code(methods/attributes) from user
* Abstraction is not much useful in python

```python
# step1 -> Importing one library (method and class)
# abc --> library name (abstract base class)
# ABC --> Class structure
# abstractmethod ---> Decorater
from abc import ABC, abstractmethod
```

```python
# step2
class Testing(ABC):
  @abstractmethod
  def Display(self):
    print('Abstraction')

class Child(Testing):
  def Display(self):
    super().Display()

```

```python
# ob = Testing()
# ob.Display()
# You can't create an object of Abstract class
ob = Child()
ob.Display()
```

# Encapsulation
* Wrapping up the data members into one single unit(Attribure +method) = capsule(class)
* We can specify which method, attributes should be passed to the derived class with the help of **Access Specifier**


---
We have three types of access specifier(In python only two):
* Public: The data that can be access anywhere(accessible to everyone)
* protected: THe data that can be accessible only in the same class and inherited class(derived class) --In python protected behave same as public
* Private: Only within the same class i can access the data

# How to use access specifier

```python
# self.name = Public attribute
# self._age = Protected attribute
# self.__salary = Private attribute (double underscore before name of variable)

# We can create method as an private attribute
# def __display(self) #-> private method
```

```python
class Person:
  def __init__(self,name,age,salary):
    self.name = name # Public
    self._age = age  # Protected
    self.__salary = salary # Private
  def display(self):
    print(f'Your name is {self.name} and your age is {self._age},Salary = {self.__salary}')
```

```python
class Employee(Person):
  def __init__(self,name,age,salary,EMPID):
    self.EMPID = EMPID
    Person.__init__(self,name,age,salary)

  def empDisplay(self):
    print(self.EMPID)
    print(self.name)
    print(self._age)
    # print(self.__salary)
```

```python
ob1 = Person('Aryan',28,1000000)
print(ob1.name)
print(ob1._age)
# print(ob1.__salary) This us my private attribute so we cannot access it outside of the class
ob1.display()
```

```python
e1 = Employee('Shivani',30,150000,200)
e1.display()
# e1.__salary Private attribute of parent class
```

```python
e1.empDisplay()
```

Create a class Rectangle

This class should have following features
* Two attributes : Length and width
* Two methods: Area and perimeter
* You are taking the lenght and widht with the help of constructot `__init__`


```python
class Rectangle:
  def __init__(self,length,width):
    self.length = length
    self.width = width

  def area(self):
    print(f'Area of Rectangle: {self.length*self.width}')

  def perimeter(self):
    print(f'Perimeter of Rectangle: {2*(self.length+self.width)}')

```

```python
ob = Rectangle(2,3)
ob.area()
ob.perimeter()
```
